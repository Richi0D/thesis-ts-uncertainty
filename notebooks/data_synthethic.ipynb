# Synthetic Data



```python
import altair as alt
import numpy as np
import polars as pl

from ts_uncertainty.plotting import plot_synthetic_sample
from ts_uncertainty.synthetic import SyntheticGenerator, SyntheticSample

alt.renderers.enable("jupyter", offline=True)
```

```python
s_gen = SyntheticGenerator(
    context_length=200,
    horizon=20,
    seasonal=True,
    covariate=False,
    noise = False,
    spikes = False,
    shift = False,
)
s = s_gen.sample(shift_point=0.5, shift_type="noise")
plot_synthetic_sample(s, title="Synthetic Sample")
```

# Simple sin wave

```python
sin_gen = SyntheticGenerator(
    context_length=96,
    horizon=24,
    seasonal=True,
    periods=(48,),             # period of the sine
    n_periods=1,
    n_harmonics=1,             # 1 harmonic = pure sine, no extra shape
    level_range=(0.0, 0.0),    # centered at 0
    slope_range=(0.0, 0.0),    # no trend
    amplitude_range=(1.0, 1.0),
    noise=False,
    spikes=False,
    shift=False,
)
s = sin_gen.sample()
plot_synthetic_sample(s, title="Sin Sample")
```
